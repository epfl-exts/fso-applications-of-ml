## Teil 2: Extrahierung von Merkmale aus den Texten

Computer verstehen natürliche Sprache in ihrer unstrukturierten Form nicht. Wie lässt sich Text also darstellen?

### „Bag of Words“ (Wörter-Sack-Modell)

Eines der einfachsten – in der Frühzeit der NLP jedoch effektiven und häufig verwendeten – Modelle zur Darstellung von Text für maschinelles Lernen ist das ***Bag-of-Words***-Modell ([Link](https://en.wikipedia.org/wiki/Bag-of-words_model)). Bei diesem Modell lassen wir den Grossteil der Struktur des Eingabetextes (Wortreihenfolge, Kapitel, Absätze, Sätze oder Formatierungen) ausser Acht und zählen lediglich, wie oft jedes Wort in einem jeweiligen Text vorkommt. Da die Struktur verworfen und nur das Vorkommen der Wörter gezählt wird, lässt sich dies bildlich als „Sack voller Wörter“ (engl. *bag of words*) veranschaulichen.

In der obigen Tabelle steht jede Spalte für ein Wort aus dem Korpus, und jede Zeile repräsentiert eines der vier Dokumente. Der Wert in jeder Zelle gibt an, wie oft das jeweilige Wort in einem bestimmten Dokument vorkommt. So tritt beispielsweise im vierten Dokument das Wort „captain“ zweimal auf, während die Wörter „my“ und „O“ jeweils einmal vorkommen.

Die technische Umsetzung des „**Bag of Words**“-Modells wird als `CountVectorizer` bezeichnet. Sie wandelt jedes Dokument in eine Zahlenreihe um – also in einen numerischen Vektor; daher auch der Name „**Vectorizer**“.

Obwohl diese Art der Transformation es Algorithmen des maschinellen Lernens ermöglicht, Textdaten effizient zu verarbeiten, hat sie einen Nachteil: Alle Wörter werden als unabhängig voneinander betrachtet, und der Kontext, in dem sie auftreten, wird ignoriert. So kann beispielsweise der Verlust von Informationen über die Wortreihenfolge die Bedeutung eines Satzes verändern. Die Sätze „I do like NLP“, „Do I like NLP“ oder „NLP like I do“ bestehen zwar aus denselben Wörtern, haben aber unterschiedliche Bedeutungen.

### TF-IDF

Der Ansatz **„Term Frequency–Inverse Document Frequency“** (TF-IDF) zielt darauf ab, diese Einschränkung zu beheben, indem er misst, wie wichtig ein Wort für ein bestimmtes Dokument im Verhältnis zu einer gesamten Dokumentensammlung (dem Korpus) ist.

Wir verwenden hierfür die Implementierung von `scikit-learn`. Der TF-IDF-Wert wird als Produkt aus folgenden Komponenten berechnet:
- Der **„Term Frequency“ (TF)**: Dies ist das Verhältnis der Häufigkeit des Wortes $w$ in einem gegebenen Dokument $d$ zur Gesamtzahl der Wörter in diesem Dokument. 
Es gilt also: $TF(w, d) = \frac{f(w, d)}{N(d)}$
- Und der (geglätteten) **„Inverse Document Frequency“ (IDF)**, die wie folgt definiert ist:
$$IDF(w, D) = \log\left(\frac{size(D)+1}{df(w, D)+1}\right)+1$$
Dabei steht $df(w, D)$ für die Anzahl der Dokumente im Korpus $D$, die das Wort $w$ enthalten. Durch das Hinzufügen von „1“ im Zähler und im Nenner bleibt der IDF-Wert endlich und stabil. Auf diese Weise erhalten häufige Wörter, die in vielen Dokumenten vorkommen (niedriger IDF-Wert), ein geringeres Gewicht, während seltene Wörter, die nur in wenigen Dokumenten auftreten, ein höheres Gewicht erhalten (hoher IDF-Wert).

### Embeddings

Die Ansätze „Bag of Words“ und TF-IDF können weder die Bedeutung von Wörtern noch die Beziehungen zwischen ihnen erfassen. Zudem führen sie zu sehr hochdimensionalen und dünnbesetzten (sparsen) Textdarstellungen, die wenig effizient sind und zu Overfitting (Überanpassung) führen können.
Um diese Einschränkungen zu überwinden, können wir **Embeddings** oder auf Transformer-Architekturen basierende Modelle einsetzen. Bei Embeddings handelt es sich um kompaktere Vektordarstellungen von Wörtern, die aus umfangreichen Textkorpora gelernt wurden. Da ähnliche Wörter durch ähnliche Vektoren repräsentiert werden, lassen sich Bedeutungen und Beziehungen in einem kontinuierlichen, niedrigdimensionalen Vektorraum abbilden.
